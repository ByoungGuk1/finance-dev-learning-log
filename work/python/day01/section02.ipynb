### 넘파이 배열의 장점 — 요소별 연산과 브로드캐스팅

파이썬 리스트에 *2를 하면 리스트가 2배로 이어붙여지지만, 넘파이 배열은 각 요소에 2를 곱합니다.  
이렇게 배열의 크기가 달라도 작은 배열을 자동으로 확장해 연산하는 것을 브로드캐스팅이라고 합니다.

````python
py_list = [1, 2, 3]
print(py_list * 2)        # [1, 2, 3, 1, 2, 3]  (리스트 반복)

np_arr = np.array([1, 2, 3])
print(np_arr * 2)         # [2 4 6]             (브로드캐스팅)
````

````python
# 3종목 x 4일 종가 (3,4)
prices = np.array([
  [71000, 71500, 70800, 72000],
  [62000, 61800, 62200, 61500],
  [155000, 154000, 156000, 153000],
])

# 날짜별 환율
fee = np.array([100,100,150,150])

prices.shape, fee.shape

# 계산하면 fee.shape가 복제되어 3열(3줄)의 2차원으로 변형

print(prices-fee)

fee_rate = np.array([[0.001],[0.0015],[0.002]])
fee_rate.shape

# 계산시에 복제 => [[a,a,a,a],[b,b,b,b],[c,c,c,c]]
print(prices * fee_rate)
````

### 요소별 연산 — 사칙연산 · 행렬곱 · 비교연산

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)   # [5 7 9]
print(a - b)   # [-3 -3 -3]
print(a * b)   # [4 10 18]  (요소별 곱셈)
print(a / b)   # [0.25 0.4 0.5]
````

````python
A = np.array([[1, 1], [0, 1]])
B = np.array([[2, 0], [3, 4]])

print(A * B)   # 요소별 곱셈
print(A @ B)   # 행렬 곱셈 (내적)
````

````python
A = np.array([1, 2, 3])
B = np.array([2])
print(A > B)   # [False False True]
````

### [LAB 3-2] 일별 수익률 계산 (벡터화 연산)   난이도 ★★★   예상 소요 25분
💰 시나리오 : 종가 배열만으로 반복문 없이 일별 수익률을 계산해야 합니다. 수익률 = (오늘종가 - 어제종가) / 어제종가  
힌트 · prices = np.array([71000, 71500, 70800, 72000, 72300, 71900])처럼 6일치 종가를 만드세요.  
힌트 · 오늘 종가는 prices[1:], 어제 종가는 prices[:-1] 슬라이싱으로 얻을 수 있습니다. 두 배열의 길이가 같아 요소별 연산이 가능합니다.  
힌트 · returns = (prices[1:] - prices[:-1]) / prices[:-1] 로 일별 수익률 배열을 구하세요.  
힌트 · np.round(returns * 100, 2)로 퍼센트(%)로 변환해 출력하세요.  
TODO : for문 없이 슬라이싱과 벡터화 연산만으로 일별 수익률 배열을 계산해 퍼센트로 출력하세요.  


[71500 70800 72000 72300 71900]  
[71000 71500 70800 72000 72300]  




````python
prices = np.array([71000, 71500, 70800, 72000, 72300, 71900])

today = prices[1:] # : index 1부터 끝까지
yesterday = prices[:-1] # : index 0<=index<끝

print(today, yesterday)

returns = (prices[1:] - prices[:-1]) / prices[:-1]

result = np.round(returns * 100, 2)

print(result)


#---

result = []
prices = [71000, 71500, 70800, 72000, 72300, 71900]

for index, data in enumerate(prices):
  if index!=0 :
    today = data
    yesterday = prices[index-1]
    returns = (today-yesterday)/yesterday
    result.append(returns)

result = [round(result*100,2) for result in result]

print(result)
````

## 배열 인덱싱과 슬라이싱

### 단일 요소 인덱싱

````python
a = np.array([0, 1, 2, 3, 4, 5])
print(a[0], a[3])

b = np.array([[0, 1, 2], [3, 4, 5]])
print(b[1, 0])   # 3
print(b[1, 2])   # 5

c = [[0, 1, 2], [3, 4, 5]]
print(c[1][2])
````

## 실전 데이터 불러오기와 결측치 처리

실무에서는 배열을 코드로 직접 입력하기보다 CSV 파일에서 불러오는 경우가 대부분입니다.  
이번 절에서는 파일에서 배열을 불러오는 방법과, 중간중간 비어 있는 값(결측치)을 처리하는 방법을 살펴봅니다.

### np.loadtxt vs np.genfromtxt

np.loadtxt는 빠르지만 빈 칸(결측치)이 있으면 오류가 납니다.  
np.genfromtxt는 조금 느리지만 빈 칸을 자동으로 nan(Not a Number)으로 처리해줘서 실무 데이터에 더 적합합니다.

````python
# stock_prices.csv 예시 (일부 결측치 포함)
# 삼성전자,SK하이닉스,카카오,NAVER,LG화학
# 71000,120000,45000,210000,410000
# 71500,,45500,209000,412000
# 70800,121000,,208500,409000

np.set_printoptions(suppress=True)

data = np.genfromtxt("stock_prices.csv", delimiter=",", skip_header=1) #, filling_values=1
print(data)          # 빈 칸은 자동으로 nan이 됩니다
print(data.dtype)    # nan을 표현하려면 float64 타입이어야 합니다

try:
  data2 = np.loadtxt("stock_prices.csv",delimiter=",",skiprows=1)
  print(data2)
  print(data2.dtype)
except ValueError as e:
  print(e)
````

💡 지수 표현을 끄는 설정 (suppress=True)  
np.set_printoptions(suppress=True)

## 결측치(NaN) 찾아서 채우기

````python
print(np.isnan(data))          # 결측치 위치를 True/False로 표시
print(np.isnan(data).sum())    # 결측치 개수

col_mean = np.nanmean(data, axis=1)        # 결측치를 제외한 열(종목)별 평균
print("col_mean : ",col_mean)
inds = np.where(np.isnan(data))
print("np.where(np.isnan(data)) : ",inds)
data[inds] = np.take(col_mean, inds[0])    # 결측치를 해당 열 평균으로 채우기

print(data)
````

주의 :  
np.nanmean, np.nansum처럼 nan으로 시작하는 함수들은 결측치를 무시하고 계산합니다.  
일반 mean()/sum()을 쓰면 결측치가 하나만 있어도 결과 전체가 nan이 되어버리니 주의하세요.

힌트 · weights = np.array([0.3, 0.25, 0.2, 0.15, 0.1])로 포트폴리오 비중을 정하고, weights @ returns.mean(axis=1)로 포트폴리오 기대수익률을 계산하세요.  
TODO : 종목별 평균 수익률/변동성, 최고·최저 수익률 종목, 포트폴리오 기대수익률까지 순서대로 계산해 출력하세요.

````python
prices = np.random.randint(50000, 200000, size=(5, 20))
returns = (prices[:, 1:] - prices[:, :-1]) / prices[:, :-1]
# print(f"월별 수익률 : {returns}")

print(f"종목별 평균 수익률 : {returns.mean(axis=1)}")
print(f"종목별 변동성 : {returns.std(axis=1)}")

argmax = np.argmax(returns.mean(axis=1))
argmin = np.argmin(returns.mean(axis=1))
print(f"argmax: {argmax}, argmin: {argmin}")

weights = np.array([0.3, 0.25, 0.2, 0.15, 0.1])
print(f"기대 수익률: {weights @ returns.mean(axis=1)}")
````
