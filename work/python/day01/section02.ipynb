### 넘파이 배열의 데이터 타입(dtype)

### 넘파이 배열의 장점 — 요소별 연산과 브로드캐스팅

파이썬 리스트에 *2를 하면 리스트가 2배로 이어붙여지지만, 넘파이 배열은 각 요소에 2를 곱합니다.  
이렇게 배열의 크기가 달라도 작은 배열을 자동으로 확장해 연산하는 것을 브로드캐스팅이라고 합니다.

````python
py_list = [1, 2, 3]
print(py_list * 2)        # [1, 2, 3, 1, 2, 3]  (리스트 반복)

np_arr = np.array([1, 2, 3])
print(np_arr * 2)         # [2 4 6]             (브로드캐스팅)
````

````python
# 3종목 x 4일 종가 (3,4)
prices = np.array([
  [71000, 71500, 70800, 72000],
  [62000, 61800, 62200, 61500],
  [155000, 154000, 156000, 153000],
])

# 날짜별 환율
fee = np.array([100,100,150,150])

prices.shape, fee.shape

# 계산하면 fee.shape가 복제되어 3열(3줄)의 2차원으로 변형

print(prices-fee)

fee_rate = np.array([[0.001],[0.0015],[0.002]])
fee_rate.shape

# 계산시에 복제 => [[a,a,a,a],[b,b,b,b],[c,c,c,c]]
print(prices * fee_rate)
````

## 넘파이 배열로 다양하게 연산하기

### 요소별 연산 — 사칙연산 · 행렬곱 · 비교연산

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(a + b)   # [5 7 9]
print(a - b)   # [-3 -3 -3]
print(a * b)   # [4 10 18]  (요소별 곱셈)
print(a / b)   # [0.25 0.4 0.5]
````

````python
A = np.array([[1, 1], [0, 1]])
B = np.array([[2, 0], [3, 4]])

print(A * B)   # 요소별 곱셈
print(A @ B)   # 행렬 곱셈 (내적)
````

````python
A = np.array([1, 2, 3])
B = np.array([2])
print(A > B)   # [False False True]
````

### 수학 함수 · 집계 함수와 벡터화 연산

### [LAB 3-2] 일별 수익률 계산 (벡터화 연산)   난이도 ★★★   예상 소요 25분
💰 시나리오 : 종가 배열만으로 반복문 없이 일별 수익률을 계산해야 합니다. 수익률 = (오늘종가 - 어제종가) / 어제종가  
힌트 · prices = np.array([71000, 71500, 70800, 72000, 72300, 71900])처럼 6일치 종가를 만드세요.  
힌트 · 오늘 종가는 prices[1:], 어제 종가는 prices[:-1] 슬라이싱으로 얻을 수 있습니다. 두 배열의 길이가 같아 요소별 연산이 가능합니다.  
힌트 · returns = (prices[1:] - prices[:-1]) / prices[:-1] 로 일별 수익률 배열을 구하세요.  
힌트 · np.round(returns * 100, 2)로 퍼센트(%)로 변환해 출력하세요.  
TODO : for문 없이 슬라이싱과 벡터화 연산만으로 일별 수익률 배열을 계산해 퍼센트로 출력하세요.  


[71500 70800 72000 72300 71900]  
[71000 71500 70800 72000 72300]  




````python
prices = np.array([71000, 71500, 70800, 72000, 72300, 71900])

today = prices[1:] # : index 1부터 끝까지
yesterday = prices[:-1] # : index 0<=index<끝

print(today, yesterday)

returns = (prices[1:] - prices[:-1]) / prices[:-1]

result = np.round(returns * 100, 2)

print(result)


#---

result = []
prices = [71000, 71500, 70800, 72000, 72300, 71900]

for index, data in enumerate(prices):
  if index!=0 :
    today = data
    yesterday = prices[index-1]
    returns = (today-yesterday)/yesterday
    result.append(returns)

result = [round(result*100,2) for result in result]

print(result)
````
