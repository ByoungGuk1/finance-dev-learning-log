💰 시나리오 : 예치금 규모에 따라 VIP/GOLD/SILVER 등급을 자동으로 판정해 창구 직원에게 안내합니다.  
힌트 · get_grade(balance) 함수를 만드세요.

힌트 · calc_interest(balance, grade) 함수도 만들어 등급별로 다른 이율(VIP 3%, GOLD 2%, SILVER 1%)로 연 이자를 계산해 반환하세요.  
TODO : get_grade와 calc_interest 두 함수를 완성하고, 예치금 3개(500만/1500만/6000만원)에 대해 등급과 예상 이자를 출력하세요.


````python
def get_grade(balance):
  if balance >= 50000000:
    return 'VIP'
  elif balance>=10000000:
      return 'GOLD'
  else:
    return 'SILVER'

def calc_interest(balance, grade):
  rate = {"VIP":0.03, "GOLD":0.02, "SILVER":0.01}
  return rate[grade] * balance

balances = [5000000, 15000000, 60000000]

for balance in balances:
  grade = get_grade(balance)
  print(grade)
  result = calc_interest(balance, grade)
  print(result)
````

# 클래스와 객체 (OOP)

- 클래스(Class) : 객체가 가지는 속성과 메소드를 정의한 설계도
- 인스턴스화(Instantiate) : 객체를 생성하는 작업, Instance : 클래스로부터 생성된 객체
- 정보은닉 : instance 변수나 메소드를 외부에서 함부로 호출하지 못하도록 하는 것 (데이터 보호가 목적)

````python
class BankEmployee:
  def __init__(self):
    pass
  def __init__(self, name, salary):
    self.name = name
    self.salary = salary

  def start_work(self):
    print(f"{self.name}이 업무 시작")
  def end_work(self):
    print(f"{self.name}이 업무 종료")
  def work(self):
    print(f"{self.name}이 업무 중")
  def get_monthly_salary(self):
    return self.salary

class Teller(BankEmployee):
  def __init__(self, name, salary):
    super().__init__(name, salary)

  def work(self):
    super().work()
    print(f"입출금 창구 업무")

class LoanOfficer(BankEmployee):
  def __init__(self, name, salary):
    super().__init__(name, salary)

  def work(self):
    super().work()
    print(f"대출 상담 및 심사")

  def get_monthly_salary(self):
    return super().get_monthly_salary()+200000

class BranchManager(BankEmployee):
  def __init__(self):
    super().__init__()
  def __init__(self, name, salary):
    super().__init__(name, salary)

  def approve_loan(self, amount):
    print(f"{amount} 만큼 대출을 승인합니다.")


employeeList = [Teller("창구직원", 100000), LoanOfficer("대출상담사",100000), BranchManager("지점장", 300000)]

for emp in employeeList:
  emp.start_work()
  emp.work()
  emp.end_work()
  # if type(emp) is BranchManager:
  #   emp.approve_loan(200000)
  if isinstance(emp, BranchManager):
    emp.approve_loan(200000)
  print(emp.get_monthly_salary())
````
