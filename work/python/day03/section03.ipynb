# Pandas

Pandas는 표(테이블) 형태의 데이터를 다루기 위한 라이브러리입니다.  
NumPy가 숫자 배열 연산에 강하다면, Pandas는 엑셀 시트처럼 행과 열로 이루어진 실제 데이터(결측치, 문자열, 날짜 혼합)를 다루는 데 최적화되어 있습니다.

## 시리즈(Series)란?

Series는 1차원 배열에 인덱스(이름표)가 붙은 자료구조입니다.

```python
import pandas as pd

scores = pd.Series([90, 85, 95], index=['국어', '영어', '수학'])
print(scores)
print(type(scores))
```

## 데이터프레임(DataFrame)이란?

DataFrame은 여러 개의 Series(열)가 모인 2차원 표 형태의 자료구조로, JSON/딕셔너리와 자연스럽게 대응됩니다.

```python
import pandas as pd

data = {
    "id": ["A001", "A002", "A003"],
    "owner": ["홍길동", "김민수", "이영희"],
    "balance": [1250000, 830000, 5200000],
}
accounts = pd.DataFrame(data)
print(accounts)
print(type(accounts))
print(accounts.owner.dtype)
```

### loc[ ] · iloc[ ]로 필터링

|구분|기준|예시|
|--|--|--|
|loc[ ]|행/열의 이름(라벨)|accounts.loc[:, ['owner','balance']]|
|iloc[ ]|행/열의 번호(정수 위치)|accounts.iloc[0:2, 1:3]|

### loc로 필터링하기

`loc`는 행·열의 이름(라벨)을 기준으로 접근합니다.

|표기|선택 대상|
|--|--|
|`df.loc[row_name, column_name]`|특정 행과 열|
|`df.loc[:, column_name]`|전체 행의 특정 열|
|`df.loc[:, [column_name_1, column_name_2]]`|전체 행의 여러 열|
|`df.loc[row_name]`|특정 행|
|`df.loc[[row_name_1, row_name_2]]`|여러 행|
|`df.loc[[row_name_1, row_name_2], [column_name_1, column_name_2]]`|여러 행과 여러 열|

### iloc로 필터링하기

`iloc`는 행·열의 정수 위치를 기준으로 접근합니다. `loc`와 비슷한 표기지만 라벨 대신 0부터 시작하는 위치를 사용합니다.

|표기|선택 대상|
|--|--|
|`df.iloc[row_index, column_index]`|특정 위치의 행과 열|
|`df.iloc[:, column_index]`|전체 행의 특정 위치 열|
|`df.iloc[:, [column_index_1, column_index_2]]`|전체 행의 여러 위치 열|
|`df.iloc[row_index]`|특정 위치 행|
|`df.iloc[[row_index_1, row_index_2]]`|여러 위치 행|
|`df.iloc[[row_index_1, row_index_2], [column_index_1, column_index_2]]`|여러 위치 행과 열|

```python
accounts.loc[:, ['owner', 'balance']]   # 전체 행, 지정한 열만
accounts.iloc[0:2, [0, 2]]               # 0~1번째 행, 0·2번째 열
```

## 5.  결측치 처리하기

실제 데이터에는 값이 비어 있는 결측치(NaN)가 자주 섞여 있습니다. 결측치를 그대로 두면 평균/합계 같은 통계 계산이 왜곡되므로 반드시 확인하고 처리해야 합니다.

### 결측치 확인하기

```python
accounts.isna().sum()     # 열별 결측치 개수
accounts.info()           # non-null 개수로도 확인 가능
```

### 결측치란?

데이터가 없거나 부적절할 때 결측치로 표시합니다.

|표현|의미|
|--|--|
|NA|Not Available, 유효한 값이 없음|
|NaN|Not a Number, 숫자가 아님|
|Null|아무 값도 없음을 의미|

결측치는 분석 결과 왜곡, 예측 모델 성능 저하, 데이터 완전성과 신뢰성 저하의 원인이 될 수 있습니다.

### 결측 비율에 따른 처리 방향

|결측 비율|제시할 수 있는 검토 방향|
|--|--|
|5% 미만|결측 행 삭제를 검토합니다. 데이터 손실과 분석 영향을 확인합니다.|
|5~20%|평균·중앙값·최빈값 대체 또는 모델 기반 처리를 검토합니다.|
|20% 이상|열 삭제 또는 대체를 검토하되, 작은 데이터셋이나 중요한 변수에서는 특히 신중히 판단합니다.|

이 비율은 절대적인 규칙이 아닙니다. 변수의 중요성, 분석 목적, 결측이 발생한 이유와 데이터 특성을 함께 고려합니다.

### 결측치 처리하기

|함수|설명|
|--|--|
|fillna(값)|결측치를 다른 값으로 대체|
|replace(old, new)|특정 값을 다른 값으로 대체|

🎤 면접에서 이렇게 말하세요  
Q. Series와 DataFrame의 차이는 무엇인가요?  
A. Series는 인덱스가 붙은 1차원 배열이고, DataFrame은 여러 개의 Series(열)가 모인 2차원 표 형태의 자료구조입니다. DataFrame에서 열 하나를 선택하면 Series가 반환된다고 예시로 설명하면 좋습니다.  
Q. loc와 iloc의 차이를 설명해보세요.  
A. loc는 행/열의 이름(라벨) 기준으로 접근하고, iloc는 행/열의 정수 위치(순번) 기준으로 접근합니다. 예를 들어 인덱스를 문자열로 바꾼 데이터프레임에서는 loc['B']는 되지만 iloc는 여전히 숫자 위치로만 접근한다는 점을 짚어주면 좋습니다.  
Q. 결측치를 처리하는 방법에는 어떤 것들이 있나요?  
A. fillna로 특정 값(평균, 기본값 등)으로 채우거나, dropna로 결측치가 있는 행/열 자체를 제거하는 방법이 있습니다. 데이터의 성격과 결측 비율에 따라 채울지 제거할지를 판단해야 한다고 답하면 실무 감각을 보여줄 수 있습니다.  
Q. groupby는 SQL의 어떤 구문과 비슷한가요?  
A. SQL의 GROUP BY와 개념이 동일합니다. 특정 열 값으로 데이터를 그룹으로 나눈 뒤 각 그룹에 집계 함수(평균, 합계 등)를 적용한다는 점에서 SELECT ... GROUP BY ... 와 대응된다고 설명하면 좋습니다.

축 스케일 설정하기

### 축 스케일

|형태|x축|y축|용도|
|--|--|--|--|
|선형|선형|선형|일정한 간격의 값 비교|
|로그-로그|로그|로그|두 축 모두 값의 범위가 넓을 때|
|로그-선형|로그|선형|x값의 범위가 넓을 때|
|선형-로그|선형|로그|지수적으로 증가하는 값 등 y값의 범위가 넓을 때|

선형 축은 같은 차이를, 로그 축은 같은 비율을 같은 간격으로 표시합니다.

```python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 생성
x = np.linspace(1, 10, 100)
y = np.exp(x)

# 선형 스케일
plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title('Linear Scale')

# 로그 스케일
plt.subplot(2, 2, 2)
plt.plot(x, y)
plt.yscale('log')
plt.title('Logarithmic Scale (y-axis)')

# 로그-선형 스케일
plt.subplot(2, 2, 3)
plt.plot(x, y)
plt.xscale('log')
plt.title('Log-Linear Scale (x-axis)')

# 선형-로그 스케일
plt.subplot(2, 2, 4)
plt.plot(x, y)
plt.yscale('log')
plt.xscale('linear')
plt.title('Linear-Logarithmic Scale')

# 서브 그래프 간격 조절
plt.tight_layout()
plt.show()
```

텍스트 추가하기

### 텍스트 추가

`text()`로 지정한 위치에 문자열을 표시할 수 있습니다. 막대 위에 값을 쓰는 용도로도 사용합니다.

enumerate(values)를 for문에 사용하여 values의 각 항목을 순회

text( ) 함수를 사용하여 그래프에 텍스트를 추가

```python
import matplotlib.pyplot as plt

categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 20]

# 막대 그래프 생성
plt.bar(categories, values, color='skyblue')

# 각 막대 머리 위치에 텍스트 추가
for i, value in enumerate(values):
  # 옵션 추가
  plt.text(i, value + 1, str(value), ha='center', va='bottom', color='black', fontsize=10)

plt.title('Bar Chart with Text Annotations')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.show()
```

선 그래프 : 객실 등급에 따른 생존율 표시하기

### 선 그래프

`plt.plot()`으로 순서에 따른 값의 변화나 추이를 표시합니다. 예제에서는 객실 등급별 생존율을 연결합니다.

```python
import pandas as pd

titanic = pd.read_csv('titanic_selena.csv')

# 객실 등급에 따른 생존자와 사망자의 평균 계산 ( 등급별 생존자합계/ 등급별 생존자건수)
pclass_survived_mean = titanic.groupby('Pclass')['Survived'].mean().reset_index()
pclass_survived_mean
```

수직 막대 그래프 : 각 승선 항구에 따른 생존자 수 확인하기

### 수직 막대 그래프

`plt.bar()`로 범주별 크기를 비교합니다. 예제에서는 각 승선 항구의 생존자 수를 표시합니다.

```python
import pandas as pd
titanic = pd.read_csv('titanic_selena.csv')

# 승선 항구에 따른 생존자의 수 계산
survived_counts = titanic[titanic['Survived'] == 1]['Embarked'].value_counts()
survived_counts
```

영역 채우기 그래프 : 나이대별 생존자와 사망자 수 표현하기

### 영역 채우기

`plt.fill_between()`으로 두 곡선 사이 또는 곡선과 기준선 사이를 채웁니다. 예제에서는 나이대별 생존자·사망자 수를 영역으로 표현합니다.

### 구간 표기 읽기

`pd.cut()`의 결과에서 `(0, 5]`는 0 초과, 5 이하를 뜻합니다. 소괄호 쪽 경계는 제외하고 대괄호 쪽 경계는 포함합니다.

```python
import pandas as pd
titanic = pd.read_csv('titanic_selena.csv')

# 결측치 처리
titanic = titanic.dropna(subset=['Age', 'Fare'])

# 나이대별 생존자와 사망자 수 계산
# 범주형 변수로 변환
age_groups = pd.cut(titanic['Age'], bins=range(0, 81, 5))

# Age, Survived 기준으로 그룹화
survived_counts = titanic.groupby([age_groups, 'Survived'], observed=False).size().unstack().fillna(0)
survived_counts
```

```python
import matplotlib.pyplot as plt

# 영역 채우기 그래프 그리기
plt.figure(figsize=(10, 6))

# 나이대별 생존자
plt.fill_between(survived_counts.index.astype(str), survived_counts[1],
                 color='purple', alpha=0.9, label='Survived')

# 나이대별 사망자
plt.fill_between(survived_counts.index.astype(str), survived_counts[0],
                 color='hotpink', alpha=0.6, label='Not Survived')

plt.title('Survival by Age Group on Titanic')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```

박스 플롯 : 승객 나이의 데이터 분포, 중앙값, 이상치 살펴보기

박스

### 박스 플롯의 박스

박스는 1사분위수 Q1부터 3사분위수 Q3까지, 가운데 50%의 범위를 나타냅니다. 박스 안의 선은 중앙값입니다.

### 사분위수와 IQR

|기준|위치|
|--|--|
|Q1|25% 지점|
|Q2|50% 지점, 중앙값|
|Q3|75% 지점|
|100% 지점|최댓값|

`IQR = Q3 - Q1`은 사분위 범위입니다. 데이터의 퍼짐과 이상치 후보를 살피는 데 사용합니다.

수염

### 박스 플롯의 수염

기본적인 1.5 × IQR 기준에서 수염은 경계 안에 있는 실제 관측값까지 뻗습니다. 경계 밖 관측값은 별도 점으로 표시합니다.

|수염|끝점|
|--|--|
|하단|Q1 − 1.5 × IQR 이상인 관측값 중 최솟값|
|상단|Q3 + 1.5 × IQR 이하인 관측값 중 최댓값|

이상치

### 이상치 후보

수염 밖의 점은 분포에서 멀리 떨어진 관측값입니다. 이상치로 표시되었다고 해서 무조건 오류나 삭제 대상인 것은 아니므로 데이터의 맥락을 확인합니다.

```python
import pandas as pd

titanic = pd.read_csv('titanic_selena.csv')
print(titanic.info(), '\n')

# 결측치 처리
titanic = titanic.dropna(subset=['Age'])
print(titanic.info())
```

```python
import matplotlib.pyplot as plt

# 승객 등급에 따른 나이의 박스 플롯
plt.boxplot([titanic[titanic['Pclass'] == 1]['Age'],
             titanic[titanic['Pclass'] == 2]['Age'],
             titanic[titanic['Pclass'] == 3]['Age']],
            labels=['1st Class', '2nd Class', '3rd Class'])

plt.title('Box Plot for Age by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Age')
plt.show()
```

### 객실 등급별 나이 분포 읽기

- 1등석: 박스와 중앙값이 상대적으로 높습니다.
- 2등석: 중앙값은 1등석보다 낮고 3등석보다 높습니다. 박스가 비교적 좁으며 수염 밖의 어린 승객도 보입니다.
- 3등석: 중앙값이 가장 낮고, 나이가 많은 쪽에 수염 밖의 점들이 보입니다.

이는 교안 그래프에서 읽을 수 있는 분포의 차이입니다. 직업이나 가족 관계까지 박스 플롯만으로 단정할 수는 없습니다.

바이올린 플롯 : 승객 등급에 따른 나이 분포 표시하기

### 바이올린 플롯

`plt.violinplot()`은 데이터 분포를 밀도 형태로 보여줍니다. 예제에서는 객실 등급별 승객 나이 분포를 비교합니다.

```python
import pandas as pd

titanic = pd.read_csv('titanic_selena.csv')
print(titanic.info(), '\n')

# 결측치 처리
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].mean())
print(titanic.info())
```

```python
import matplotlib.pyplot as plt

# 바이올린 플롯 그리기
plt.figure(figsize=(10, 6))

# titanic[‘Pclass’] == i는 Pclass는 열이 i인 승객들을 선택하고, 그중에서 Age 열만 추출
# showmeans=False는 평균값을 표시하지 않도록 하고, showmedians=True는 중앙값을 표시하도록 함
violin_plot = plt.violinplot([titanic[titanic['Pclass'] == 1]['Age'],
                              titanic[titanic['Pclass'] == 2]['Age'],
                              titanic[titanic['Pclass'] == 3]['Age']],
                             showmeans=False, showmedians=True)

plt.title('Violin Plot of Age by Pclass')
plt.xlabel('Pclass')
plt.ylabel('Age')

# x축의 눈금 설정
plt.xticks([1, 2, 3], ['1st Class', '2nd Class', '3rd Class'])

# 범례 설정
plt.legend(violin_plot['bodies'], ['1st Class', '2nd Class', '3rd Class'], title='Pclass')
plt.show()
```

에러 바 : 요금의 평균과 표준편차 표현하기

### 에러 바

`plt.errorbar()`로 대표값과 함께 변동성이나 불확실성 범위를 표시합니다. 이 예제는 부모·자녀 수별 운임의 평균과 **표준편차**를 사용합니다.

에러 바가 나타내는 데이터 신뢰성은 다음과 같은 요소를 포함

### 에러 바 해석

|표시하는 값|해석|
|--|--|
|측정 오차|측정값의 불확실성|
|표준편차|관측값이 평균 주위로 퍼진 정도|
|표준오차|추정 평균의 변동성|
|신뢰구간|추정에 사용한 절차의 신뢰수준에 따른 구간|

에러 바가 무엇을 나타내는지 먼저 확인해야 합니다. 이 교안 예제의 에러 바는 표준편차이며, 신뢰구간이 아닙니다. 두 에러 바의 겹침만으로 통계적 유의성을 단정하지 않습니다.

```python
import pandas as pd
titanic = pd.read_csv('titanic_selena.csv')

# 각 부모와 자녀의 수에 대한 요금의 평균과 표준 편차 계산
# 평균
fare_means = titanic.groupby('Parch')['Fare'].mean()
print(fare_means, '\n')

# 표준 편차
fare_std = titanic.groupby('Parch')['Fare'].std()
print(fare_std)
```

에러바는 데이터의 표준 편차를 나타내는 경우에는 길이가 길수록 해당 그룹의 데이터가 퍼져 있음을 의미

```python
import matplotlib.pyplot as plt

# 에러바로 요금의 평균과 표준 편차 표현
plt.figure(figsize=(10, 6))

# 에러바 생성
plt.errorbar(fare_means.index, fare_means, yerr=fare_std, fmt='o',
             capsize=5, capthick=1, label='Fare')

plt.title('Error Bar Plot of Fare by Parch')
plt.xlabel('Parch')
plt.ylabel('Fare')
plt.xticks(fare_means.index)
plt.legend()
plt.show()
```

타이타닉 데이터셋으로 개별 서브플롯 하나씩 그리기

### 타이타닉 데이터의 서브플롯

부모·자녀 수별 사망자와 생존자 수를 집계해, 한 Figure의 위쪽에는 사망자 선 그래프를, 아래쪽에는 생존자 막대 그래프를 그립니다.

## 정리 - 데이터 시각화 라이브러리, Matplotlib

p56 titanic data를 이용한 data분석에 대해 실습진행하고
실습후 분석된 결과를 정리하여 제출한다. (예, 어떤 특성을 가진 승객이 생존율이 높았다.......)

### 1. 분석 목적 및 데이터 확인

Titanic 승객 데이터를 이용하여 성별, 객실 등급, 승선 항구, 나이, 운임, 동반 가족 규모에 따른 생존율 차이를 분석한다. `Survived`는 `0`이면 사망, `1`이면 생존을 의미하므로 그룹별 평균을 계산하면 생존율을 구할 수 있다.

전체 데이터는 891명이며 `Age` 177개, `Cabin` 687개, `Embarked` 2개의 결측치가 있다. 나이별 분석에서는 `Age` 결측치를 제외한 714명을 사용한다. `Cabin`은 결측치가 많아 이번 분석 대상에서 제외한다.

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

titanic = pd.read_csv('titanic_selena.csv')

basic_summary = pd.DataFrame({
    '항목': ['전체 승객', '생존자', '사망자', '전체 생존율(%)'],
    '값': [
        len(titanic),
        int(titanic['Survived'].sum()),
        int((titanic['Survived'] == 0).sum()),
        round(titanic['Survived'].mean() * 100, 1)
    ]
})

missing_summary = titanic.isna().sum().rename('결측치 수').to_frame()
missing_summary['결측 비율(%)'] = (
    missing_summary['결측치 수'] / len(titanic) * 100
).round(1)

display(basic_summary)
display(missing_summary[missing_summary['결측치 수'] > 0])
```

### 2. 승객 특성별 생존율 표

그룹마다 탑승 인원이 다르기 때문에 생존자 수만 비교하면 결과를 잘못 해석할 수 있다. 아래 코드에서는 각 그룹의 전체 인원, 생존자, 사망자와 생존율을 함께 계산한다.

```python
def survival_summary(data, group_column):
    summary = (
        data.groupby(group_column, observed=True)['Survived']
        .agg(['count', 'sum', 'mean'])
        .rename(columns={
            'count': '전체 인원',
            'sum': '생존자',
            'mean': '생존율(%)'
        })
    )
    summary['사망자'] = summary['전체 인원'] - summary['생존자']
    summary['생존율(%)'] = (summary['생존율(%)'] * 100).round(1)
    return summary[['전체 인원', '생존자', '사망자', '생존율(%)']]

sex_summary = survival_summary(titanic, 'Sex')
pclass_summary = survival_summary(titanic, 'Pclass')
embarked_summary = survival_summary(titanic.dropna(subset=['Embarked']), 'Embarked')

age_data = titanic.dropna(subset=['Age']).copy()
age_data['AgeGroup'] = pd.cut(
    age_data['Age'],
    bins=[0, 10, 20, 40, 60, float('inf')],
    labels=['0-9', '10-19', '20-39', '40-59', '60+'],
    right=False
)
age_summary = survival_summary(age_data, 'AgeGroup')

fare_data = titanic.copy()
fare_data['FareGroup'] = pd.qcut(
    fare_data['Fare'],
    q=4,
    labels=['Low', 'Mid-Low', 'Mid-High', 'High']
)
fare_summary = survival_summary(fare_data, 'FareGroup')

family_data = titanic.copy()
family_data['FamilySize'] = family_data['SibSp'] + family_data['Parch'] + 1
family_data['FamilyGroup'] = pd.cut(
    family_data['FamilySize'],
    bins=[0, 1, 4, float('inf')],
    labels=['Alone', '2-4', '5+']
)
family_summary = survival_summary(family_data, 'FamilyGroup')

for title, summary in [
    ('성별 생존율', sex_summary),
    ('객실 등급별 생존율', pclass_summary),
    ('승선 항구별 생존율', embarked_summary),
    ('나이대별 생존율', age_summary),
    ('운임 구간별 생존율', fare_summary),
    ('가족 규모별 생존율', family_summary)
]:
    print(title)
    display(summary)
```

### 3. Matplotlib을 이용한 생존율 시각화

각 특성의 생존율을 동일한 0~100% 범위에서 비교한다. 그래프 제목과 축 이름은 실행 환경의 한글 폰트 경고를 피하기 위해 영문으로 작성한다.

```python
def draw_survival_rate(ax, summary, title, color):
    labels = summary.index.astype(str)
    rates = summary['생존율(%)']
    bars = ax.bar(labels, rates, color=color, edgecolor='black', alpha=0.8)
    ax.bar_label(bars, fmt='%.1f%%', padding=3)
    ax.set_title(title)
    ax.set_xlabel('Group')
    ax.set_ylabel('Survival Rate (%)')
    ax.set_ylim(0, 100)
    ax.grid(axis='y', linestyle='--', alpha=0.4)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

draw_survival_rate(axes[0, 0], sex_summary, 'Survival Rate by Sex', 'salmon')
draw_survival_rate(axes[0, 1], pclass_summary, 'Survival Rate by Pclass', 'royalblue')
draw_survival_rate(axes[0, 2], embarked_summary, 'Survival Rate by Embarked Port', 'mediumorchid')
draw_survival_rate(axes[1, 0], age_summary, 'Survival Rate by Age Group', 'seagreen')
draw_survival_rate(axes[1, 1], fare_summary, 'Survival Rate by Fare Group', 'goldenrod')
draw_survival_rate(axes[1, 2], family_summary, 'Survival Rate by Family Size', 'slateblue')

fig.suptitle('Titanic Passenger Survival Analysis', fontsize=18, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

### 4. 분석 결과 및 결론

전체 승객 891명 중 생존자는 342명으로 전체 생존율은 **38.4%**였다.

- **성별:** 여성의 생존율은 74.2%, 남성은 18.9%로 여성의 생존율이 뚜렷하게 높았다.
- **객실 등급:** 1등석은 63.0%, 2등석은 47.3%, 3등석은 24.2%로 객실 등급이 높을수록 생존율이 높았다.
- **승선 항구:** 생존자 수는 Southampton이 가장 많았지만, 전체 인원을 고려한 생존율은 Cherbourg가 55.4%로 가장 높았다. Queenstown은 39.0%, Southampton은 33.7%였다.
- **나이:** 0~9세의 생존율은 61.3%로 높았고, 60세 이상은 26.9%로 낮았다. 어린이가 구조 과정에서 우선 보호되었을 가능성이 있다.
- **운임:** 운임 하위 25% 구간의 생존율은 19.7%, 상위 25% 구간은 58.1%였다. 높은 운임은 높은 객실 등급과 관련되어 있어 생존율 차이에도 함께 영향을 주었을 가능성이 있다.
- **가족 규모:** 혼자 탑승한 승객은 30.4%, 2~4명 가족은 57.9%, 5명 이상 가족은 16.1%였다. 적당한 규모의 가족과 함께 탑승한 승객의 생존율이 가장 높았다.

종합하면 **여성, 1등석, 높은 운임, 0~9세, 2~4명 가족과 동반한 승객**의 생존율이 상대적으로 높았다. 다만 이 분석은 변수 사이의 관련성을 확인한 것이며, 특정 변수가 생존의 직접적인 원인임을 증명하는 것은 아니다. 성별, 객실 등급, 운임, 나이와 가족 구성이 복합적으로 생존 결과와 관련된 것으로 해석할 수 있다.
